## **Data Cleaning**

```python
import numpy as np

print("--- Data Cleaning Examples ---")

# 1. drop() - Drop a column
df_after_drop = df.copy()
df_after_drop = df_after_drop.drop(columns=['size'])
print("\nDataFrame after dropping 'size' column:")
print(df_after_drop.head())

# 2. dropna() and 3. fillna() - Handle missing values
df_with_nans = df.copy()
# Introduce some NaN values for demonstration
df_with_nans.loc[0, 'tip'] = np.nan
df_with_nans.loc[1, 'sex'] = np.nan
df_with_nans.loc[2, 'day'] = np.nan

print("\nDataFrame with introduced NaNs:")
print(df_with_nans.head())

# dropna()
df_after_dropna = df_with_nans.dropna()
print("\nDataFrame after dropna():")
print(df_after_dropna.head())
print(f"Original rows: {len(df_with_nans)}, Rows after dropna: {len(df_after_dropna)}")

# fillna()
# Fill 'tip' with its mean, 'sex' with 'Unknown', and 'day' with 'Other'
df_after_fillna = df_with_nans.fillna({'tip': df_with_nans['tip'].mean(),
                                       'sex': 'Unknown',
                                       'day': 'Other'})
print("\nDataFrame after fillna():")
print(df_after_fillna.head())

# 4. replace() - Replace specific values
df_after_replace = df.copy()
# Replace 'Female' with 'Woman' in the 'sex' column
df_after_replace['sex'] = df_after_replace['sex'].replace('Female', 'Woman')
print("\n'sex' column after replacing 'Female' with 'Woman':")
print(df_after_replace['sex'].value_counts())

# 5. str.lower() - Convert string column to lowercase
df_after_lower = df.copy()
# Convert 'day' column to lowercase
df_after_lower['day'] = df_after_lower['day'].str.lower()
print("\n'day' column after converting to lowercase:")
print(df_after_lower['day'].value_counts())

# 6. astype() - Change data type
df_after_astype = df.copy()
# Convert 'tip' column from float to int
# Note: This will truncate decimal values.
df_after_astype['tip'] = df_after_astype['tip'].astype(int)
print("\nDataFrame head and info after converting 'tip' to int:")
print(df_after_astype.head())
print(df_after_astype.info())

# 7. drop_duplicates() - Remove duplicate rows
# Create a sample DataFrame with duplicates for demonstration
data_with_duplicates = {
    'col1': [1, 2, 1, 3, 2, 4],
    'col2': ['A', 'B', 'A', 'C', 'B', 'D'],
    'col3': [10, 20, 10, 30, 20, 40]
}
df_duplicates_example = pd.DataFrame(data_with_duplicates)
print("\nDataFrame with duplicates for drop_duplicates example:")
print(df_duplicates_example)

df_no_duplicates = df_duplicates_example.drop_duplicates()
print("\nDataFrame after drop_duplicates():")
print(df_no_duplicates)

print("\nDataFrame after drop_duplicates(subset=['col1']):")
df_no_duplicates_subset = df_duplicates_example.drop_duplicates(subset=['col1'])
print(df_no_duplicates_subset)
```
